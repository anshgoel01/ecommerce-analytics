```python
import sqlite3
from pathlib import Path

import pandas as pd
import plotly.express as px

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DB_PATH = PROJECT_ROOT / "data" / "processed" / "olist.db"
conn = sqlite3.connect(DB_PATH)

def q(sql):
    return pd.read_sql(sql, conn)
```

## 1. Monthly revenue trend by category

```python
rev = q("SELECT * FROM v_monthly_revenue_by_category ORDER BY order_month")
rev.head()
```

```python
top_categories = (
    rev.groupby("category")["gmv"].sum().sort_values(ascending=False).head(10).index
)
fig = px.line(
    rev[rev["category"].isin(top_categories)],
    x="order_month", y="gmv", color="category",
    title="Monthly GMV — top 10 categories by total revenue",
)
fig.show()
```

```python
# Top 5 growing / declining categories: compare the average of each
# category's LAST 3 months of GMV vs its FIRST 3 months, rather than a
# single first-vs-last month. A single-month comparison blows up here
# because the marketplace itself launched small (Sep 2016 had ~2 orders
# total across the whole platform) -- a category going from one R$73
# order to R$41K reads as 56,000% "growth" when it is really just launch
# ramp-up, not a trend. Also require a minimum GMV floor in the first
# window so a near-zero denominator cannot produce a similarly misleading
# swing in the other direction.
MIN_FIRST_WINDOW_GMV = 300

span = rev.groupby("category")["order_month"].nunique()
eligible = span[span >= 6].index

def first_last_avg(g):
    g = g.sort_values("order_month")
    return pd.Series({
        "first_3mo_avg_gmv": g["gmv"].head(3).mean(),
        "last_3mo_avg_gmv": g["gmv"].tail(3).mean(),
    })

windows = (
    rev[rev["category"].isin(eligible)]
    .groupby("category")
    .apply(first_last_avg, include_groups=False)
)
windows = windows[windows["first_3mo_avg_gmv"] >= MIN_FIRST_WINDOW_GMV]
windows["pct_change"] = (
    100 * (windows["last_3mo_avg_gmv"] - windows["first_3mo_avg_gmv"]) / windows["first_3mo_avg_gmv"]
)

print("Top 5 growing:")
display(windows.sort_values("pct_change", ascending=False).head(5))
print("Top 5 declining:")
display(windows.sort_values("pct_change").head(5))
```

## 2. Seller performance segmentation

```python
sellers = q("SELECT * FROM v_seller_performance")
fig = px.scatter(
    sellers, x="revenue", y="avg_review_score", color="quadrant",
    size="orders", hover_data=["seller_id", "late_delivery_rate"],
    title="Seller segmentation: revenue vs. satisfaction",
)
fig.show()
sellers["quadrant"].value_counts()
```

## 3. Delivery delay vs. review score

```python
delay = q("SELECT * FROM v_delivery_delay_vs_review")
fig = px.bar(delay, x="delay_bucket", y="avg_review_score", text="orders",
             title="Average review score by delivery-delay bucket")
fig.show()
```
